# scBaseCamp metadata

```python
# list all records for a feature in scbasecamp_metadata
feat_type = "GeneFull_Ex50pAS"
#feat_type = "Velocyto"
#feat_type = "Gene"
tbl = Table("scbasecamp_metadata")
stmt = Query \
    .from_(tbl) \
    .select("*") \
    .where(tbl.feature_type == feat_type)

with db_connect() as conn:
    df = pd.read_sql(str(stmt), conn)
    print(df)
```
